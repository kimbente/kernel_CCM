```python
import torch
```

```python
# Load all data
# Links are changed because of nested structure
co2_vostok = torch.load("../data/vostok/co2/CO2_vostok_400kyr_timeseries.pt")
# co2_epica = torch.load("data/epica/co2/CO2_epica_800kyr_timeseries.pt")

insol_vostok = torch.load("../data/vostok/insolation/insolation_65N_400kyr_timeseries.pt")
# insol_epica = torch.load("data/epica/insolation/insolation_65N_800kyr_timeseries.pt")

ch4_vostok = torch.load("../data/vostok/ch4/CH4_vostok_400kyr_timeseries.pt")
# ch4_epica = torch.load("data/epica/ch4/CH4_epica_800kyr_timeseries.pt")

temp_vostok = torch.load("../data/vostok/temp/TEMP_vostok_400kyr_timeseries.pt")
# temp_epica = torch.load("data/epica/temp/TEMP_epica_800kyr_timeseries.pt")
```

```python
co2_vostok_stan = (co2_vostok - co2_vostok.mean()) / co2_vostok.std()
insol_vostok_stan = (insol_vostok - insol_vostok.mean()) / insol_vostok.std()
ch4_vostok_stan = (ch4_vostok - ch4_vostok.mean()) / ch4_vostok.std()
temp_vostok_stan = (temp_vostok - temp_vostok.mean()) / temp_vostok.std()
```

```python
torch.save(co2_vostok_stan, "../data/vostok/co2/CO2_vostok_stan_400kyr_timeseries.pt")
torch.save(insol_vostok_stan, "../data/vostok/insolation/insolation_65N_stan_400kyr_timeseries.pt")
torch.save(ch4_vostok_stan, "../data/vostok/ch4/CH4_vostok_stan_400kyr_timeseries.pt")
torch.save(temp_vostok_stan, "../data/vostok/temp/TEMP_vostok_stan_400kyr_timeseries.pt")
```
